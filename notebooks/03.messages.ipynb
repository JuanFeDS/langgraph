### **📩 Manejo de mensajes**

```python
# Mensaje de IA
ai_msg = AIMessage(
    content='Message from AI'
)
ai_msg
```

```python
# Mensaje humano
human_msg = HumanMessage(
    content='Message from human'
)
human_msg
```

```python
# Opción para manejo del historial del chat
history_chat = [human_msg, ai_msg]
```

```python
# Visualización del historial de chat
for message in history_chat:
    message.pretty_print()
```
