### **📚 Librerías**

```python
import os
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model
```

### **🤖 Creación de un RAG**

```python
# Variables de entorno
rag_id = os.getenv("OPENAI_RAG_ID")
rag_papeleria = os.getenv("OPENAI_RAG_ID_PAPELERIA")
```

```python
# Inicialización del LLM
llm = init_chat_model(
    model = "openai:gpt-4o-mini", 
    temperature=0
)
```

```python
# Configuración de tool nativa de langchain
file_search_tool = {
    "type": "file_search",
    "vector_store_ids": [rag_papeleria]
}
```

```python
# Agregamos la tool al LLM
llm_with_tools = llm.bind_tools([file_search_tool])
```

```python
# Realizamos la llamada
response = llm_with_tools.invoke("""
    Me gustaría llevar 10 cuadernos, 4 boligrafos azules, 
    1 corrector liquido y 5 papeles bond carta, 
    cuanto me costaria?
""")
```

```python
print(response.text)
```
